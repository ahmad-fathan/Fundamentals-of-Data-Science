## A.1 Python dalam data science

Pekerjaan data science umumnya mengikuti alur yang berulang: mengumpulkan data, membersihkan
dan menyiapkannya, mengeksplorasi dengan statistik dan grafik, memodelkan, lalu
mengomunikasikan hasil. Python dipakai di seluruh tahapan tersebut karena sintaksnya mudah
dibaca, komunitasnya besar, dan terutama karena ekosistem library (kumpulan kode siap pakai)

## A.2 Notebook: Jupyter, Google Colab, dan VS Code

Data science umumnya dikerjakan di **notebook**, yaitu dokumen interaktif yang menggabungkan
kode, hasil eksekusi, grafik, dan teks penjelasan. Berbeda dengan file script `.py` yang
dijalankan sekaligus dari awal sampai akhir, file notebook `.ipynb` terdiri atas **cell** yang
dapat dijalankan satu per satu, sehingga hasil setiap langkah langsung terlihat. Ada dua jenis
cell:

- **Code cell** berisi kode Python, dan hasilnya muncul tepat di bawah cell.
- **Text cell (Markdown cell)** berisi penjelasan yang ditulis dengan Markdown, format teks
  sederhana seperti `# Judul` atau `**tebal**`.

Kode di notebook dijalankan oleh **runtime** (istilah Colab) atau **kernel** (istilah Jupyter
dan VS Code), yaitu proses Python aktif yang menyimpan semua variabel dan library yang sudah
diimpor selama sesi berjalan.

Tiga lingkungan notebook yang umum dipakai adalah:

- **Jupyter Notebook**, lingkungan notebook interaktif berbasis browser yang dipasang di
  komputer sendiri.
- **Google Colab**, layanan notebook Jupyter yang berjalan di server Google. Tidak
  memerlukan instalasi, library utama sudah tersedia, dan notebook disimpan di Google Drive.
- **VS Code**, editor kode yang mendukung script maupun notebook melalui ekstensi Python dan
  Jupyter.

Modul ini memakai **Google Colab** sebagai lingkungan praktik utama karena hasilnya sama di
semua komputer tanpa perlu instalasi.

## A.5 Mengganti jenis runtime

**Runtime** adalah mesin virtual di server Google yang menjalankan kode notebook. Secara
bawaan, runtime memakai **CPU**, dan itu sudah cukup untuk seluruh praktik di modul ini. Untuk
pekerjaan berat seperti melatih model deep learning, runtime dapat diganti agar memakai
akselerator:

1. Pilih menu **Runtime > Change runtime type**.
2. Pada **Runtime type**, biarkan `Python 3`.
3. Pada **Hardware accelerator**, pilih `CPU`, `GPU` (misalnya T4 GPU), atau `TPU`. Sebagian
   pilihan hanya tersedia untuk pengguna Colab berbayar.
4. Klik **Save**.

Mengganti jenis runtime memulai sesi baru, sehingga semua variabel hilang dan seluruh cell perlu
dijalankan ulang. Kuota GPU gratis juga terbatas, jadi pakai akselerator hanya bila memang
diperlukan.

memunculkan tombol **Choose Files**:

Untuk menyalin path file yang tepat, klik kanan file di panel **Files** lalu pilih **Copy path**.

### A.6.2 Mengakses file di Google Drive

Berbeda dengan penyimpanan sesi, Google Drive bersifat **permanen**. Drive dihubungkan (mount)
ke runtime dengan ikon **Mount Drive** di panel Files, atau dengan kode berikut. Saat dijalankan,
Colab meminta izin akses ke akun Google:

````python
# Hanya berjalan di Google Colab
try:
    from google.colab import drive
    import pandas as pd

    drive.mount("/content/drive")
    df_drive = pd.read_csv("/content/drive/MyDrive/data/nilai.csv")
    print(df_drive.head())
except ImportError:
    print("Cell ini hanya berjalan di Google Colab (modul google.colab tidak tersedia).")
    print('Di lokal, ganti path menjadi: df_drive = pd.read_csv("data/nilai.csv")')
except Exception as galat:
    print("Mount Google Drive dibatalkan atau gagal:", galat)
````

Folder `MyDrive` adalah folder utama Google Drive pengguna, sehingga path di atas merujuk ke
file `nilai.csv` di dalam folder `data`. Ganti path tersebut sesuai lokasi file masing-masing.
Cara ini cocok untuk dataset yang dipakai berulang kali, karena tidak perlu diunggah ulang
setiap sesi.

## A.7 Urutan eksekusi cell

Hasil notebook bergantung pada **urutan cell dijalankan**, bukan urutan cell ditulis. Variabel
yang dibuat di sebuah cell disimpan di runtime dan baru dapat dipakai cell lain setelah cell
pembuatnya dijalankan.

**Contoh kasus.** Sebuah notebook berisi dua cell. Apa yang terjadi bila cell kedua dijalankan
lebih dulu?

````python
# Cell 1: membuat variabel
nilai = 80

# Cell 2: memakai variabel dari Cell 1
print(nilai + 5)
````

**Hasil yang diharapkan bila Cell 1 dijalankan lebih dulu.**

```text
85
```

**Interpretasi.** Bila Cell 2 dijalankan sebelum Cell 1, atau setelah runtime dimulai ulang,
Python menampilkan `NameError` karena `nilai` belum ada di runtime. Angka di dalam kurung siku
di kiri cell, misalnya `[3]`, menunjukkan urutan eksekusinya. Masalah serupa muncul bila sebuah
cell diubah tetapi tidak dijalankan ulang, karena runtime masih menyimpan nilai lama.

Kesalahan jenis ini sering tidak memunculkan pesan galat sama sekali, sehingga sulit dikenali.
Untuk memastikan notebook benar dari awal, pilih **Runtime > Restart session and run all**.
Perintah ini menghapus semua variabel di runtime, lalu menjalankan seluruh cell dari atas ke
bawah. Lakukan langkah ini sebelum mengumpulkan tugas.

## A.8 Menggunakan Gemini di Colab

Colab memiliki asisten AI bawaan bernama **Gemini**. Gemini memahami isi notebook yang sedang
dibuka, termasuk kode dan variabel yang sudah dijalankan (**Gambar 6** pada modul):

1. Ikon Gemini (bintang berkilau) di bagian bawah notebook. Klik ikon ini untuk membuka atau
   menutup panel Gemini.
2. Tab **Gemini**, yaitu panel percakapan yang terbuka di sisi kanan notebook.
3. Contoh pertanyaan yang dapat diklik langsung, misalnya cara memasang library atau memuat data
   dari Google Drive.
4. Kotak perintah (prompt), tempat menuliskan pertanyaan atau permintaan dalam bahasa
   sehari-hari, termasuk bahasa Indonesia.
5. Pemilih model, untuk memilih versi model Gemini yang dipakai.
6. Tautan **generate** pada code cell kosong, untuk meminta Gemini menuliskan kode langsung di
   cell tersebut.

Beberapa cara memakai Gemini yang berguna untuk praktik di modul ini:

- **Menjelaskan kode.** Contoh perintah: *"Jelaskan apa yang dilakukan cell ini baris demi baris."*
- **Membuat kode.** Contoh perintah: *"Buat histogram kolom nilai dari DataFrame df dengan judul
  dan label sumbu."*
- **Menjelaskan dan memperbaiki galat.** Saat sebuah cell menghasilkan galat, klik tombol
  **Explain error** di bawah pesan galat.
- **Menganalisis data secara bertahap.** *Data Science Agent* dapat menyusun rencana analisis,
  menjalankan kode, dan merangkum temuannya.

Usulan perubahan kode ditampilkan dalam bentuk perbandingan (diff) yang dapat diterima atau
ditolak. Kode dari Gemini tetap perlu dibaca dan diuji sebelum dipakai, karena hasilnya bisa
keliru atau tidak sesuai konteks soal. Untuk latihan dan tugas, gunakan Gemini sebagai **tutor
yang menjelaskan**, bukan sebagai pengganti proses berpikir sendiri, dan ikuti ketentuan dosen
tentang penggunaan AI.
